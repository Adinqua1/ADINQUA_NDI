# Exploratory Data Analysis: NTCDA-Amine Condensation & $\pi$-Stacking

## 1. Data Source & Access Description

### Dataset Overview: Laboratory Experimental Reactions
* **Data Type:** Primary laboratory experimental dataset (Tabular CSV/Excel format).
* **Source / Provider:** Primary experimental data collected in-house during organic synthesis trials.
* **Access Method:** Stored locally in the repository as `DATA/ntcda_amine_yields.csv` (or imported directly via `pandas.read_csv()`).
* **Variables Measured:**
  * **Amine Type (Categorical):** Primary amines varied by alkyl chain length, steric bulk, or aromaticity (e.g., Hexylamine, Octylamine, Aniline, Phenethylamine).
  * **Reaction Temperature (°C) (Continuous Numerical):** Controlled thermal reaction conditions ranging from 120°C to 200°C.
  * **Reaction Time (hours) (Continuous Numerical):** Duration of reflux/heating phase.
  * **Percentage Yield (%) (Continuous Numerical):** Isolated mass yield relative to theoretical maximum yield.
  * **$\pi$-Stacking Interplanar Distance ($\text{\AA}$) (Continuous Numerical):** Characterization metric derived via Powder X-Ray Diffraction (PXRD).

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style
sns.set_theme(style="whitegrid")

# --- Generate Sample Experimental Data (Replace with your actual pd.read_csv('...')) ---
np.random.seed(42)
amines = ['Hexylamine', 'Octylamine', 'Aniline', 'Phenethylamine']
temperatures = [120, 140, 160, 180, 200]  # in °C

data_list = []
for amine in amines:
    for temp in temperatures:
        for trial in range(3): # 3 replicate trials per condition
            # Base yield simulation with temperature effect
            base_yield = 40 + (temp - 120) * 0.25 if 'ylamine' in amine else 30 + (temp - 120) * 0.15
            yield_val = np.clip(base_yield + np.random.normal(0, 4), 10, 98)
            pi_dist = 3.8 - (temp - 120) * 0.002 + np.random.normal(0, 0.05)
            
            data_list.append({
                'Amine': amine,
                'Temperature_degC': temp,
                'Trial': trial + 1,
                'Yield_percent': round(yield_val, 2),
                'Pi_Stacking_Distance_A': round(pi_dist, 3)
            })

df = pd.DataFrame(data_list)

# --- Figure 1: Reaction Yield vs Temperature across Amines ---
plt.figure(figsize=(9, 5))
sns.lineplot(
    data=df, 
    x='Temperature_degC', 
    y='Yield_percent', 
    hue='Amine', 
    style='Amine',
    markers=True, 
    dashes=False, 
    err_style='bars', 
    ci=68, 
    palette='Set2'
)

plt.title("Effect of Reaction Temperature on NDI Percentage Yield by Amine Type", fontsize=12, fontweight='bold')
plt.xlabel("Reaction Temperature (°C)", fontsize=11)
plt.ylabel("Percentage Yield (%)", fontsize=11)
plt.ylim(0, 100)
plt.xticks(temperatures)
plt.legend(title="Amine Substrate", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()

# --- Figure 2: Experimental Parameter Coverage Matrix ---
plt.figure(figsize=(8, 4))
coverage_matrix = df.groupby(['Amine', 'Temperature_degC']).size().unstack(fill_value=0)
sns.heatmap(coverage_matrix, annot=True, cmap="YlGnBu", cbar_kws={'label': 'Number of Experimental Replicates'})

plt.title("Experimental Parameter Coverage Matrix", fontsize=12, fontweight='bold')
plt.xlabel("Reaction Temperature (°C)", fontsize=11)
plt.ylabel("Amine Substrate", fontsize=11)
plt.tight_layout()
plt.show()
```
